```python
from misa.constants import DATABASE_PATH
from misa.dataaccess import CommentsDataSet
from misa.preprocess import MentionsRemover, PunctuationRemover, StopWordsRemover, WordStemmer
from time import monotonic
```

```python
dataset = CommentsDataSet(DATABASE_PATH)
df = dataset.dataset("dev-00000-of-00001").raw()
df.head()
```

```python
preprocessors = (MentionsRemover(), PunctuationRemover(), StopWordsRemover())
# , WordStemmer()
comments = df["text"].tolist()
len(comments)
```

```python
transformed = [*comments]
for preprocessor in preprocessors:
    transformed = [preprocessor.run(comment) for comment in transformed]    
```

```python
import demoji

res = {}
for comment in transformed:
    res = dict(res, **demoji.findall(comment))

len(res)
```
